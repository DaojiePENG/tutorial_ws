## 2.1 Python 简介

Python 是一种高级编程语言，结合了解释性、编译性、互动性和面向对象的特性。它以简洁易读的语法设计而闻名，非常适合初学者和专业开发者。

Python 的主要特点包括：

* 解释型语言：无需编译，代码可以直接运行，类似于 PHP 和 Perl。

* 交互式：支持在终端输入代码并立即执行，方便测试和调试。

* 面向对象：支持面向对象编程，允许代码封装在对象中。

* 易学易用：语法简单，关键字少，代码结构清晰。

* 跨平台：支持在 UNIX、Windows 和 Mac 等多个平台运行。

* 丰富的标准库：内置大量模块，支持文件操作、网络编程、数据库访问等功能。

* 可扩展性：可以通过 C 或 C++ 编写高性能模块，并在 Python 中调用。

* 支持 GUI：可以开发图形用户界面应用程序。

* 嵌入性：可以嵌入到其他语言（如 C/C++）的程序中，提供脚本化能力。

Python 由 Guido van Rossum 于 20 世纪 80 年代末设计，受多种语言（如 ABC、C、SmallTalk 等）启发。它的开源特性和强大的社区支持使其成为全球最流行的编程语言之一，广泛应用于数据科学、人工智能、Web 开发、自动化等领域。

## 2.2 使用SDK —— 导入依赖包

在编写机器人控制程序之前，我们需要先导入所需的Python库。就像学外语前要准备好词典一样，我们需要把SDK中的工具"请"进来。

```python
import time  # 时间模块，用于控制程序等待/延时
import sys   # 系统模块，用于访问系统相关功能

# 从宇树SDK中导入通信通道相关模块
from unitree_sdk2py.core.channel import ChannelSubscriber, ChannelFactoryInitialize

# 从宇树SDK中导入运动控制客户端
from unitree_sdk2py.go2.sport.sport_client import (
    SportClient,           # 运动控制客户端，用于发送运动指令
    PathPoint,             # 路径点，用于定义运动轨迹
    SPORT_PATH_POINT_SIZE, # 路径点数组的大小常量
)
```

### 2.2.1 初始化运动控制客户端

初始化通信通道并创建运动控制客户端。这一步相当于"拨通电话"，建立你的电脑与机器狗之间的连接。

```python
# 初始化通信通道，参数0表示域ID，"ens37"是网络接口名称
ChannelFactoryInitialize(0, "ens37")

# 创建运动控制客户端实例
sport_client = SportClient()

# 设置通信超时时间为10秒（如果10秒内没有响应则报错）
sport_client.SetTimeout(10.0)

# 初始化客户端，完成与机器狗的连接
sport_client.Init()
```

### 2.2.2 尝试站立动作

调用 `StandUp()` 方法让机器狗站起来。返回值为0表示指令执行成功。

```python
# 发送站立指令，机器狗将从当前姿态站起来
ret = sport_client.StandUp()
print("返回值: ", ret)  # 返回0表示成功
```

### 2.2.3 尝试趴下动作

调用 `StandDown()` 方法让机器狗趴下。这是站立的反向操作。

```python
# 发送趴下指令，机器狗将缓慢趴下
ret = sport_client.StandDown()
print("返回值: ", ret)  # 返回0表示成功
```

### 2.2.4 尝试坐下与起立

调用 `Sit()` 让机器狗坐下，等待2秒后再调用 `RiseSit()` 让它从坐姿起立。通过 `time.sleep()` 控制动作间隔。

```python
# 发送坐下指令
ret = sport_client.Sit()

# 等待2秒，让机器狗完成坐下动作
time.sleep(2)

# 发送从坐姿起立的指令
ret = sport_client.RiseSit()
```
