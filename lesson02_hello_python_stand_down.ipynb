### 2.3.4 函数定义

把常用的操作封装成可重用的函数。


````python
def stand_and_wait(sport_client, duration=2):
    """
    让机器人站立并等待指定时间
    
    参数：
        sport_client: 运动控制客户端
        duration: 等待时间（秒）
    
    返回值：
        True if successful, False otherwise
    """
    try:
        ret = sport_client.StandUp()
        if ret == 0:
            print(f"✓ 站立成功，等待 {duration} 秒")
            time.sleep(duration)
            return True
        else:
            print(f"✗ 站立失败，返回码: {ret}")
            return False
    except Exception as e:
        print(f"✗ 错误: {e}")
        return False

# 使用函数
stand_and_wait(sport_client, duration=3)
````

### 2.3.5 异常处理（try-except）

程序遇到错误时不会崩溃，而是优雅地处理。


````python
try:
    # 可能出错的代码
    ret = sport_client.StandUp()
    if ret != 0:
        raise Exception(f"站立命令返回错误码: {ret}")
    
    time.sleep(2)
    
except ConnectionError:
    # 处理连接错误
    print("❌ 与Go2连接断开，请检查网络")
    
except TimeoutError:
    # 处理超时错误
    print("⏱ 命令超时，Go2未在规定时间内响应")
    
except Exception as e:
    # 处理其他所有错误
    print(f"❌ 发生错误: {e}")
    
finally:
    # 无论是否出错都会执行
    print("✓ 操作完成（或失败）")
````

## 2.4 Go2 的运动指令详解

### 2.4.1 核心运动命令

| 命令 | 作用 | 参数 | 返回值 |
|------|------|------|--------|
| `StandUp()` | 从任何姿态站起 | 无 | 0=成功, 非0=失败 |
| `StandDown()` | 趴下 | 无 | 0=成功, 非0=失败 |
| `Sit()` | 坐下 | 无 | 0=成功, 非0=失败 |
| `RiseSit()` | 从坐姿站起 | 无 | 0=成功, 非0=失败 |
| `Move(vx, vy, vyaw)` | 持续运动 | vx(前后), vy(左右), vyaw(旋转) | 持续执行 |
| `Stop()` | 立即停止 | 无 | 0=成功 |

**返回值说明：**
- `0`：命令执行成功
- `非0`：命令执行失败（具体错误码根据SDK文档）


### 2.4.2 运动指令的详细使用

**站立和趴下：**

````python
# 站立
ret = sport_client.StandUp()
if ret == 0:
    print("✓ 站立成功")
else:
    print(f"✗ 站立失败，错误码: {ret}")

# 趴下
ret = sport_client.StandDown()
if ret == 0:
    print("✓ 趴下成功")
time.sleep(2)  # 等待完成

# 坐下和起立
ret = sport_client.Sit()
time.sleep(1)
ret = sport_client.RiseSit()
````

**连续运动（Move命令）：**

````python
# Move(vx, vy, vyaw) 参数说明：
# vx：前后速度（正值向前，负值向后，单位 m/s）
# vy：左右速度（正值向左，负值向右，单位 m/s）
# vyaw：旋转速度（正值逆时针，负值顺时针，单位 rad/s）

# 向前走
for i in range(10):
    sport_client.Move(0.5, 0, 0)  # 0.5 m/s 向前
    time.sleep(0.1)

# 原地转圈
for i in range(20):
    sport_client.Move(0, 0, 0.5)  # 0.5 rad/s 逆时针
    time.sleep(0.1)

# 侧向移动
for i in range(15):
    sport_client.Move(0, 0.3, 0)  # 0.3 m/s 向左
    time.sleep(0.1)

# 复合动作：斜向前进
for i in range(20):
    sport_client.Move(0.3, 0.2, 0.1)  # 向前、向左、旋转
    time.sleep(0.1)
````

## 2.5 实践项目：编写第一个完整程序

### 2.5.1 简单示例：站立和趴下

创建文件 `simple_stand_down.py`：

````python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

"""
简单的站立和趴下演示

这个程序展示了如何：
1. 连接到Go2
2. 让Go2站立
3. 让Go2趴下
4. 处理基本的错误
"""

import time
import sys

from unitree_sdk2py.core.channel import ChannelFactoryInitialize
from unitree_sdk2py.go2.sport.sport_client import SportClient


def main():
    """主函数"""
    
    print("="*60)
    print("【Go2 站立和趴下演示】")
    print("="*60)
    
    try:
        # 初始化
        print("\n[初始化] 正在连接到 Go2...")
        ChannelFactoryInitialize(0, "ens37")
        sport_client = SportClient()
        sport_client.SetTimeout(10.0)
        sport_client.Init()
        print("✓ 连接成功！")
        
        # 演示站立
        print("\n[动作1] 执行站立...")
        ret = sport_client.StandUp()
        if ret == 0:
            print("✓ Go2 已站立")
            time.sleep(2)  # 让机器人保持站立状态2秒
        else:
            print(f"✗ 站立失败，错误码: {ret}")
            return False
        
        # 演示趴下
        print("\n[动作2] 执行趴下...")
        ret = sport_client.StandDown()
        if ret == 0:
            print("✓ Go2 已趴下")
            time.sleep(2)
        else:
            print(f"✗ 趴下失败，错误码: {ret}")
            return False
        
        # 再次站立
        print("\n[动作3] 再次站立...")
        ret = sport_client.StandUp()
        if ret == 0:
            print("✓ Go2 已再次站立")
            time.sleep(1)
        else:
            print(f"✗ 站立失败，错误码: {ret}")
            return False
        
        print("\n" + "="*60)
        print("✓ 演示完成！所有动作均成功执行")
        print("="*60)
        return True
        
    except Exception as e:
        print(f"\n❌ 发生错误: {e}")
        import traceback
        traceback.print_exc()  # 打印详细的错误堆栈
        return False


if __name__ == "__main__":
    success = main()
    sys.exit(0 if success else 1)
````

### 2.5.2 进阶示例：参数化函数与完整流程

创建文件 `advanced_choreography.py`：

````python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

"""
进阶示例：参数化函数与动作组合

演示如何：
1. 创建可复用的函数
2. 组合多个动作
3. 添加错误处理和重试逻辑
"""

import time
import sys

from unitree_sdk2py.core.channel import ChannelFactoryInitialize
from unitree_sdk2py.go2.sport.sport_client import SportClient


class Go2Controller:
    """Go2机器人控制器类"""
    
    def __init__(self, nic_name="ens37", timeout=10.0):
        """
        初始化机器人控制器
        
        参数：
            nic_name: 网络接口名称
            timeout: 命令超时时间（秒）
        """
        self.nic_name = nic_name
        self.timeout = timeout
        self.sport_client = None
        self.action_count = 0  # 统计动作数量
        
    def connect(self):
        """连接到机器人"""
        print(f"[连接] 使用网卡 {self.nic_name} 连接到 Go2...")
        try:
            ChannelFactoryInitialize(0, self.nic_name)
            self.sport_client = SportClient()
            self.sport_client.SetTimeout(self.timeout)
            self.sport_client.Init()
            print("✓ 连接成功！")
            return True
        except Exception as e:
            print(f"✗ 连接失败: {e}")
            return False
    
    def execute_action(self, action_name, action_func, wait_time=1):
        """
        执行一个动作
        
        参数：
            action_name: 动作名称（用于打印）
            action_func: 动作函数（如 self.sport_client.StandUp）
            wait_time: 动作后等待时间（秒）
        
        返回值：
            True if successful, False otherwise
        """
        if self.sport_client is None:
            print("✗ 还未连接到机器人！")
            return False
        
        print(f"\n[动作 {self.action_count+1}] {action_name}...")
        try:
            ret = action_func()
            if ret == 0:
                print(f"✓ {action_name} 成功")
                time.sleep(wait_time)
                self.action_count += 1
                return True
            else:
                print(f"✗ {action_name} 失败，错误码: {ret}")
                return False
        except Exception as e:
            print(f"✗ {action_name} 出错: {e}")
            return False
    
    def perform_routine(self, routine_name, routine_func):
        """
        执行一个完整的动作序列
        
        参数：
            routine_name: 序列名称
            routine_func: 包含动作序列的函数
        """
        print(f"\n{'='*60}")
        print(f"【{routine_name}】")
        print(f"{'='*60}")
        
        try:
            routine_func()
            print(f"\n✓ {routine_name} 完成，总动作数: {self.action_count}")
            return True
        except Exception as e:
            print(f"\n✗ {routine_name} 失败: {e}")
            return False


# 定义各种动作序列
def stand_sit_routine(controller):
    """站立-坐下-起立 的动作序列"""
    controller.execute_action("站立", controller.sport_client.StandUp, wait_time=1)
    controller.execute_action("坐下", controller.sport_client.Sit, wait_time=1)
    controller.execute_action("起立", controller.sport_client.RiseSit, wait_time=1)


def stand_down_routine(controller):
    """站立-趴下-站立 的动作序列"""
    controller.execute_action("站立", controller.sport_client.StandUp, wait_time=1)
    time.sleep(2)
    controller.execute_action("趴下", controller.sport_client.StandDown, wait_time=2)
    time.sleep(1)
    controller.execute_action("站立", controller.sport_client.StandUp, wait_time=1)


def multiple_stands(controller, count=3):
    """重复站立和趴下"""
    for i in range(count):
        print(f"\n--- 循环 {i+1}/{count} ---")
        controller.execute_action("站立", controller.sport_client.StandUp, wait_time=1)
        controller.execute_action("趴下", controller.sport_client.StandDown, wait_time=1)


def continuous_movement(controller, duration=5):
    """连续运动演示"""
    print("\n[动作] 连续向前运动 5 秒...")
    try:
        start_time = time.time()
        while time.time() - start_time < duration:
            controller.sport_client.Move(0.3, 0, 0)  # 0.3 m/s 向前
            time.sleep(0.1)
        
        controller.sport_client.Stop()  # 停止
        print("✓ 连续运动完成")
        controller.action_count += 1
    except Exception as e:
        print(f"✗ 连续运动失败: {e}")


def main():
    """主程序"""
    
    # 创建控制器
    controller = Go2Controller(nic_name="ens37", timeout=10.0)
    
    # 连接
    if not controller.connect():
        print("无法连接到机器人，程序退出")
        return False
    
    print("\n" + "="*60)
    print("准备执行多个动作序列")
    print("="*60)
    
    # 执行不同的动作序列
    try:
        # 序列1：站立-坐下-起立
        controller.perform_routine("站立-坐下-起立序列", 
                                   lambda: stand_sit_routine(controller))
        
        time.sleep(2)
        
        # 序列2：站立-趴下-站立
        controller.perform_routine("站立-趴下-站立序列", 
                                   lambda: stand_down_routine(controller))
        
        time.sleep(2)
        
        # 序列3：重复站立和趴下（3次）
        controller.perform_routine("重复站立和趴下（3次）", 
                                   lambda: multiple_stands(controller, count=3))
        
        # 最终统计
        print("\n" + "="*60)
        print(f"✓ 所有演示完成！")
        print(f"总计执行了 {controller.action_count} 个动作")
        print("="*60)
        
        return True
        
    except KeyboardInterrupt:
        print("\n\n用户中断")
        return False
    except Exception as e:
        print(f"\n❌ 程序出错: {e}")
        import traceback
        traceback.print_exc()
        return False


if __name__ == "__main__":
    success = main()
    sys.exit(0 if success else 1)
````

**3. 检查返回值**

```python
# 不好的方式（忽视返回值）
sport_client.StandUp()
sport_client.StandDown()

# 好的方式（检查返回值）
ret = sport_client.StandUp()
if ret != 0:
    print(f"错误: StandUp返回{ret}")
    # 处理错误
```

### ⚠️ 常见陷阱

❌ **不要这样做：**
```python
# 没有检查返回值
sport_client.StandUp()

# 动作执行太快，机器人来不及反应
sport_client.StandUp()
sport_client.StandDown()  # 立即执行

# 不处理异常
sport_client.Init()  # 如果连接失败程序会崩溃

# 不同时期多个运动指令冲突
sport_client.Move(0.5, 0, 0)
sport_client.StandUp()  # 冲突！
```

✅ **应该这样做：**
```python
# 检查返回值
ret = sport_client.StandUp()
if ret != 0:
    print(f"错误: {ret}")

# 动作间添加延迟
sport_client.StandUp()
time.sleep(1)  # 等待完成
sport_client.StandDown()

# 用try-except处理异常
try:
    sport_client.Init()
except Exception as e:
    print(f"连接失败: {e}")

# 动作指令不冲突
sport_client.Move(0.5, 0, 0)
for i in range(10):
    time.sleep(0.1)  # 持续运动
sport_client.Stop()
time.sleep(0.5)
sport_client.StandUp()  # 现在可以执行了
```
